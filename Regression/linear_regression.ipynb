```python
import os
import numpy as np
```

```python
X = np.array([[1,3], [2, 1], [3, 2]])
y = np.array([4, 3, 6])
```

```python
weights = np.zeros(shape=X[0].shape)
bias = 0
lr = 0.05
n = X.shape[0]
```

```python
y_pred = X @ weights + bias
y_pred
```

```python
errors = (y_pred - y)**2 
errors
```

```python
loss = (1/(2*n)) * np.sum(errors)
loss
```

```python
epoch = 8
loss_showcase = []
for j in range(epoch):
    for i in range(weights.size):
        gradient = (1/n) * sum((y_pred - y) * X[:, i])
        weights[i] -= (lr * gradient)

    gradient_b = ((1/n) * sum(y_pred - y))
    bias -= (lr * gradient_b)
    
    y_pred = X @ weights + bias
    loss = (1/n) * sum((y_pred - y)**2)
    loss_showcase.append(loss)
    print(f"In the epoch: {j}, the loss is {loss}")
print(f"The updated weights are {weights} and bias is {bias}")
```

```python
import matplotlib.pyplot as plt
```

```python
x_showcase = np.arange(0, epoch, step=1)
plt.plot(x_showcase, loss_showcase)
plt.show()
```
