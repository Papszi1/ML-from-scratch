## Instead of using the already programmed gradient descend to find the optimal weights and bias, I'm using nudging.

### Start with randomly defined weights and bias, and for each iteration, add a small number to them, and check how it's influences the loss function. Based on that information, update the weights

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import importlib
import linear_regression
```

```python
X = np.array([[1,3], [2, 1], [3, 2]])
y = np.array([4, 3, 6])
```

```python
l = X.shape[0]
n = X.shape[1]
step_size = 10e-3
weights = np.zeros(n)
bias = 0
y_pred = X @ weights + bias
epoch = 100
```

```python
loss_showcase = []
y_pred = X @ weights + bias
current_best_loss = (1/l) * np.sum((y_pred - y)**2)
loss_showcase.append(current_best_loss)
```

```python
for i in range(epoch):
    for j in range(n):
        weights[j] += step_size
        y_pred = X @ weights + bias
        loss_pos = (1/l) * np.sum((y_pred - y)**2)
        
        if loss_pos < current_best_loss:
            current_best_loss = loss_pos
        else:
            weights[j] -= 2 * step_size
            y_pred = X @ weights + bias
            loss_neg = (1/l) * np.sum((y_pred - y)**2)
            
            if loss_neg < current_best_loss:
                current_best_loss = loss_neg
            else:
                weights[j] += step_size
                
    loss_showcase.append(current_best_loss)
```

```python
x_showcase = np.arange(0, epoch+1, step = 1)
plt.plot(x_showcase, loss_showcase)
plt.show()
```
